# 00 · Environment Check

Verifies that the core scientific stack is installed. Run it with the kernel
**Python 3 (EMBL-EBI DS .venv)** (create it with `make setup`; see RUNNING.md).

```python
import importlib
import logging
import sys

logging.disable(logging.INFO)  # silence DEBUG logs from mgnipy's dependencies

print(f"Python {sys.version.split()[0]}")
packages = ["numpy", "pandas", "matplotlib", "seaborn", "sklearn", "torch", "Bio", "requests", "mgnipy", "snakemake"]
for name in packages:
    try:
        module = importlib.import_module(name)
        print(f"{name:<12} {getattr(module, '__version__', 'installed')}")
    except ImportError:
        print(f"{name:<12} NOT INSTALLED")
```
