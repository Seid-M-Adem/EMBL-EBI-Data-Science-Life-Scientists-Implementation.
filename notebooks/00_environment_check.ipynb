# 00 · Environment Check

Verifies that the core scientific stack from `requirements.txt` is installed.

```python
import importlib
import sys

print(f"Python {sys.version.split()[0]}")
packages = ["numpy", "pandas", "matplotlib", "seaborn", "sklearn", "torch", "Bio", "requests"]
for name in packages:
    try:
        module = importlib.import_module(name)
        print(f"{name:<12} {getattr(module, '__version__', 'installed')}")
    except ImportError:
        print(f"{name:<12} NOT INSTALLED")
```
